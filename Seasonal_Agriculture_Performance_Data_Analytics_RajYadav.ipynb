## 9. Outlier Analysis

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)

outlier_summary
```

## 12. Seasonal Comparison


```python
# General seasonal summary

season_metrics = df.groupby("Season").agg(
    Records=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Profit=("Profit_INR", "mean"),
    Total_Profit=("Profit_INR", "sum"),
    Average_Water_Used=("Water_Used_m3", "mean")
).round(2)

season_metrics
```

```python
# Compare crop performance within each season

crop_season_summary = (
    df.groupby(["Season", "Crop"])
      .agg(
          Average_Yield=("Yield_Tonnes_Ha", "mean"),
          Average_Profit=("Profit_INR", "mean"),
          Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
      )
      .round(2)
      .sort_values(["Season", "Average_Yield"], ascending=[True, False])
)

crop_season_summary
```

## 13. Additional Analysis


```python
# Analysis 1 — Crop-specific seasonal performance
# Question:How does crop yield vary across different seasons?

crop_season_yield = df.groupby(
    ['Crop', 'Season']
)['Yield_Tonnes_Ha'].mean().unstack()

print("Average Yield (Tonnes/Ha) by Crop and Season:")
display(crop_season_yield.round(2))

# Visualization
plt.figure(figsize=(12, 6))
crop_season_yield.plot(kind='bar', figsize=(12, 6))

plt.title("Average Crop Yield Across Seasons")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.legend(title="Season")
plt.tight_layout()
plt.show()
# Interpretation:
# This analysis shows that crop performance changes across seasons.
# For example, Sugarcane has substantially higher average yield than the other crops, while most crops show their highest average yield during Kharif and lower yield during Zaid.
```

```python
# Analysis 2 — Irrigation method and water efficiency
# Question:How does the irrigation method relate to water usage, water efficiency and crop yield?
irrigation_analysis = df.groupby('Irrigation_Method').agg(
    Average_Water_Used_m3=('Water_Used_m3', 'mean'),
    Average_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean'),
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Number_of_Farms=('Farm_ID', 'count')
)

print("Irrigation Method Analysis:")
display(irrigation_analysis.round(2))

# Visualization
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df,
    x='Irrigation_Method',
    y='Water_Efficiency_t_per_1000m3'
)

plt.title("Water Efficiency by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency (Tonnes per 1000 m³)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# Interpretation:
# Rainfed farms have the highest average water-efficiency value in this dataset, while flood irrigation has the lowest.
# Drip irrigation shows a higher average yield than flood irrigation, but it also has higher average water usage.
# These results describe associations in the dataset and do not establish that an irrigation method alone causes higher or lower performance.
```

```python
# Analysis 3 — Regional economic performance
# Question:How does agricultural profitability differ across states?

state_analysis = df.groupby('State').agg(
    Average_Profit_INR=('Profit_INR', 'mean'),
    Average_Revenue_INR=('Revenue_INR', 'mean'),
    Average_Cost_INR=('Total_Cost_INR', 'mean'),
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Number_of_Farms=('Farm_ID', 'count')
)

state_analysis['Profit_Margin_%'] = (
    state_analysis['Average_Profit_INR'] /
    state_analysis['Average_Revenue_INR']
) * 100

state_analysis = state_analysis.sort_values(
    'Average_Profit_INR',
    ascending=False
)

print("State-wise Economic Performance:")
display(state_analysis.round(2))

# Visualization
plt.figure(figsize=(12, 6))
sns.barplot(
    x=state_analysis.index,
    y=state_analysis['Average_Profit_INR']
)

plt.title("Average Agricultural Profit by State")
plt.xlabel("State")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Interpretation:
# The analysis indicates differences in average profitability across states.
# Punjab and Maharashtra have the highest average profit in the dataset, whereas Andhra Pradesh has the lowest average profit among the listed states.
```

## 14. Key Insights



**1. Kharif shows the highest average yield among the three seasons.**  
- **Observation:** Kharif has an average yield of approximately 5.64 tonnes/ha, compared with 5.08 tonnes/ha in Rabi and 4.67 tonnes/ha in Zaid.
- **Evidence:** Seasonal yield comparison.
- **Interpretation:** Agricultural performance in terms of average yield is strongest in the Kharif records.
- **Limitation:** The analysis shows an association with season and does not prove that season alone caused the difference.

**2. Kharif also shows the highest average profit.**  
- **Observation:** Average profit is approximately ₹178,915 in Kharif, compared with ₹87,689 in Rabi and a negative average profit of approximately ₹24,805 in Zaid.
- **Evidence:** Seasonal economic analysis.
- **Interpretation:** Kharif records have the strongest average economic performance in the dataset.
- **Limitation:** Profit is influenced by multiple factors such as cost, production and market price.

**3. Crop performance varies considerably across seasons.**  
- **Observation:** Most crops show higher average yields in Kharif and lower yields in Zaid.
- **Evidence:** Crop-season yield analysis.
- **Interpretation:** Seasonal conditions are associated with different crop performance levels.
- **Limitation:** The dataset does not establish a causal relationship between season and crop yield.

**4. Sugarcane has much higher yield values than the other crops.**  
- **Observation:** Sugarcane records have substantially higher average yield values than crops such as Wheat, Rice, Maize and Pulses.
- **Evidence:** Crop-specific yield comparison.
- **Interpretation:** Crop type is an important factor when comparing raw yield values.
- **Limitation:** Yield should not be compared without considering crop characteristics and production requirements.

**5. Rainfed farms have the highest average water-efficiency value.**  
- **Observation:** Rainfed farms show an average water-efficiency value of approximately 7.56 tonnes per 1000 m³.
- **Evidence:** Irrigation-method analysis.
- **Interpretation:** Rainfed records use less recorded irrigation water and consequently show higher water efficiency in this dataset.
- **Limitation:** Rainfall contribution is not equivalent to zero water requirement, so this result should not be interpreted as proof that rainfed farming is always more efficient.

**6. Flood irrigation has the lowest average water efficiency among the listed irrigation methods.**  
- **Observation:** Flood irrigation has an average water-efficiency value of approximately 3.44 tonnes per 1000 m³.
- **Evidence:** Irrigation-method comparison.
- **Interpretation:** Flood irrigation records show comparatively lower water-use efficiency.
- **Limitation:** Farm conditions, crop selection and environmental factors may also contribute to this difference.

**7. Drip irrigation has a higher average yield than flood irrigation in the dataset.**  
- **Observation:** Drip-irrigated farms have an average yield of approximately 6.62 tonnes/ha compared with approximately 4.90 tonnes/ha for flood irrigation.
- **Evidence:** Irrigation and yield analysis.
- **Interpretation:** Drip irrigation is associated with higher average yield in these records.
- **Limitation:** This is an observational comparison and does not establish that irrigation method alone caused the higher yield.

**8. Profitability differs substantially across states.**  
- **Observation:** Punjab and Maharashtra have the highest average profits, while Andhra Pradesh has the lowest average profit among the states in the dataset.
- **Evidence:** State-wise economic analysis.
- **Interpretation:** Geographic location is associated with differences in agricultural economic performance.
- **Limitation:** State-level differences may reflect variations in crops, costs, market prices and farming conditions.

**9. Yield has a strong positive association with water efficiency.**  
- **Observation:** The correlation between Yield_Tonnes_Ha and Water_Efficiency_t_per_1000m3 is approximately 0.915.
- **Evidence:** Correlation analysis.
- **Interpretation:** Higher yield values tend to occur alongside higher water-efficiency values in this dataset.
- **Limitation:** Correlation does not imply causation and may partly reflect how the efficiency measure is constructed.

**10. The dataset contains some missing values.**  
- **Observation:** The dataset contains 120 missing values across its variables.
- **Evidence:** Missing-value analysis.
- **Interpretation:** Data quality should be considered when interpreting the results.
- **Limitation:** The effect of missing values depends on which variables and records contain them.

## 15. Recommendations

**1. Consider seasonal differences when planning crop activities.**  
Since Kharif records show the highest average yield and profit, seasonal performance should be considered when planning agricultural activities. However, crop-specific and regional conditions should also be evaluated.

**2. Evaluate irrigation practices with both yield and water efficiency.**  
The analysis shows considerable differences between irrigation methods. Farmers and planners should evaluate irrigation methods using both crop yield and water-efficiency measures rather than considering water consumption alone.

**3. Give attention to flood-irrigation efficiency.**  
Flood irrigation has the lowest average water-efficiency value among the analyzed irrigation methods. Where technically and economically appropriate, alternative irrigation approaches could be investigated.

**4. Analyze crop-season combinations before making crop decisions.**  
Because crop yields vary across seasons, agricultural planning should consider the interaction between crop type and season instead of evaluating crops independently.

**5. Consider regional economic differences.**  
The substantial variation in average profit between states suggests that regional factors should be included in agricultural planning and economic analysis.
